# 02 - Archetype assignment (local test)

Thin runner over `core.archetypes` and `app.archetype_assignment`.
Reference tables live in `config/archetypes/`; this notebook only calls
and looks at the result.

```python
import geopandas as gpd
import pandas as pd
from shapely.geometry import box

from city_information_modeler.app.archetype_assignment import (
    assign_archetypes,
    load_references,
)
from city_information_modeler.parallel import Runner

references = load_references()
print(list(references))
print(references["periods"][["period_id", "period_label", "istat_census_code"]])
```

```python
rows = [
    {"usage": "residential", "construction_year": 1932, "construction_year_from": 1919,
     "construction_year_to": 1945, "n_floors": 4, "height_m": 13.0, "n_dwellings": 12},
    {"usage": "residential", "construction_year": 1978, "construction_year_from": 1971,
     "construction_year_to": 1980, "n_floors": 1, "height_m": 6.0, "n_dwellings": 1},
    {"usage": "residential", "construction_year": None, "construction_year_from": None,
     "construction_year_to": None, "n_floors": 6, "height_m": 18.0, "n_dwellings": None},
]
geometries = [box(7.64 + i * 0.002, 45.02, 7.641 + i * 0.002, 45.021) for i in range(3)]
buildings = gpd.GeoDataFrame(pd.DataFrame(rows), geometry=geometries, crs="EPSG:4326")
result = assign_archetypes(buildings, runner=Runner(profile_name="sequential"))
print(result[["archetype_period_id", "archetype_class", "archetype_class_source",
              "tabula_archetype", "u_wall_primary", "window_u_mean"]])
```
