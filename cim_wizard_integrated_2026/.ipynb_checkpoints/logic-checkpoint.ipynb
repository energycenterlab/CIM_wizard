# CIM Wizard Integrated — logic (step 1)

This notebook is a **step-by-step ipynb** companion to `cim_wizard_integrated`: start here with **reading a project scenario** from the **dockerized PostgreSQL/PostGIS** database and loading it into a **GeoDataFrame**.

- **Table:** `cim_vector.cim_wizard_project_scenario` (same as SQLAlchemy `ProjectScenario` in `app/models/vector.py`).
- **Docker:** from `cim_wizard_integrated_2026/`, `docker compose up -d db` — DB is published on **localhost:15432** (see `docker-compose.yml`). Inside Compose, the service hostname is `db` on port `5432`.
- **No FastAPI required:** only SQLAlchemy + GeoPandas + a reachable DB URL.

A parallel walkthrough lives under `notebooks/01_project_scenario_to_geodataframe.ipynb`.

## 1 — Connect to the database

Set `DATABASE_URL`, or use defaults matching `docker-compose.yml`: user `cim_wizard_user`, password `cim_wizard_password`, database `cim_wizard_integrated`, host `localhost`, port `15432`.

```python
import os
from sqlalchemy import create_engine, text
from sqlalchemy.exc import OperationalError

_pg_user = os.environ.get("PGUSER", "cim_wizard_user")
_pg_pass = os.environ.get("PGPASSWORD", "cim_wizard_password")
_pg_db = os.environ.get("PGDATABASE", "cim_wizard_integrated")
_pg_host = os.environ.get("CIM_DB_HOST", "localhost")
_pg_port = os.environ.get("CIM_DB_PORT", "15432")

if os.environ.get("DATABASE_URL"):
    DATABASE_URL = os.environ["DATABASE_URL"]
else:
    DATABASE_URL = (
        f"postgresql://{_pg_user}:{_pg_pass}@{_pg_host}:{_pg_port}/{_pg_db}"
    )

engine = create_engine(DATABASE_URL)

_safe = DATABASE_URL
if "@" in _safe:
    _safe = _safe.split("@", 1)[-1]
print("Will connect to:", _safe)

try:
    with engine.connect() as _conn:
        _conn.execute(text("SELECT 1"))
    print("OK: database reachable from this kernel.")
except OperationalError as e:
    raise ConnectionError(
        "Cannot reach PostgreSQL. From this kernel, 'localhost' is THIS machine running Python, "
        "not necessarily the host where Docker runs.\n\n"
        "Fix: start the DB where this kernel can see it (e.g. `docker compose up -d db` on the same host), "
        "or set DATABASE_URL / CIM_DB_HOST to a reachable host, or use an SSH tunnel to forward port 15432.\n"
    ) from e
```

## 2 — List scenarios (pick ids)

Same data the API exposes via `CimWizardDataManager.list_projects` / `get_scenarios_for_project` (`app/core/data_manager.py`).

```python
import pandas as pd

list_sql = text("""
SELECT project_id, scenario_id, project_name, scenario_name,
       project_zoom, project_crs,
       (project_boundary IS NOT NULL) AS has_boundary,
       (project_center IS NOT NULL) AS has_center,
       (census_boundary IS NOT NULL) AS has_census_boundary,
       created_at
FROM cim_vector.cim_wizard_project_scenario
ORDER BY updated_at DESC NULLS LAST
LIMIT 50
""")

with engine.connect() as conn:
    scenarios_df = pd.read_sql(list_sql, conn)

scenarios_df
```

### Optional — all scenarios for one `project_id`

Set `LIST_PROJECT_ID` or env `CIM_LIST_PROJECT_ID`. If empty, the first row from the table above is used.

```python
LIST_PROJECT_ID = os.environ.get("CIM_LIST_PROJECT_ID")

if not LIST_PROJECT_ID:
    try:
        if len(scenarios_df) > 0:
            LIST_PROJECT_ID = scenarios_df.iloc[0]["project_id"]
    except NameError:
        pass

if not LIST_PROJECT_ID:
    raise ValueError(
        "Set LIST_PROJECT_ID or CIM_LIST_PROJECT_ID, or run the list cell above first."
    )

project_list_sql = text("""
SELECT project_id, scenario_id, project_name, scenario_name,
       project_zoom, project_crs,
       (project_boundary IS NOT NULL) AS has_boundary,
       (project_center IS NOT NULL) AS has_center,
       (census_boundary IS NOT NULL) AS has_census_boundary,
       created_at, updated_at
FROM cim_vector.cim_wizard_project_scenario
WHERE project_id = :pid
ORDER BY scenario_name NULLS LAST, updated_at DESC NULLS LAST
""")

with engine.connect() as conn:
    project_scenarios_df = pd.read_sql(
        project_list_sql, conn, params={"pid": LIST_PROJECT_ID}
    )

print("project_id:", LIST_PROJECT_ID, "→", len(project_scenarios_df), "scenario(s)")
project_scenarios_df
```

## 3 — Choose `PROJECT_ID` and `SCENARIO_ID`

Use env `CIM_PROJECT_ID` / `CIM_SCENARIO_ID`, or the first row from the listing cells above.

```python
PROJECT_ID = os.environ.get("CIM_PROJECT_ID") or None
SCENARIO_ID = os.environ.get("CIM_SCENARIO_ID") or None

try:
    _listed = scenarios_df
except NameError:
    _listed = None
try:
    _per_project = project_scenarios_df
except NameError:
    _per_project = None

if (not PROJECT_ID or not SCENARIO_ID) and _listed is not None and len(_listed) > 0:
    PROJECT_ID = _listed.iloc[0]["project_id"]
    SCENARIO_ID = _listed.iloc[0]["scenario_id"]
    print("Using first scenario from recent list:", PROJECT_ID, SCENARIO_ID)
elif (not PROJECT_ID or not SCENARIO_ID) and _per_project is not None and len(_per_project) > 0:
    PROJECT_ID = _per_project.iloc[0]["project_id"]
    SCENARIO_ID = _per_project.iloc[0]["scenario_id"]
    print("Using first scenario for project list:", PROJECT_ID, SCENARIO_ID)

if not PROJECT_ID or not SCENARIO_ID:
    raise ValueError(
        "Set CIM_PROJECT_ID / CIM_SCENARIO_ID, or run the listing cells so a row can be used."
    )
```

## 4 — Load one scenario into a GeoDataFrame

GeoPandas uses **one** active geometry column. Here **`project_boundary`** is mapped to `geometry` (typical for mapping). `project_center` and `census_boundary` are returned as WKT strings for optional parsing in the next cell.

```python
import geopandas as gpd
from shapely import wkt as shapely_wkt

scenario_sql = text("""
SELECT
  project_id,
  scenario_id,
  project_name,
  scenario_name,
  project_zoom,
  project_crs,
  grid_id,
  created_at,
  updated_at,
  project_boundary::geometry AS geometry,
  ST_AsText(project_center) AS project_center_wkt,
  ST_AsText(census_boundary) AS census_boundary_wkt
FROM cim_vector.cim_wizard_project_scenario
WHERE project_id = :pid AND scenario_id = :sid
""")

gdf = gpd.read_postgis(
    scenario_sql,
    engine,
    geom_col="geometry",
    params={"pid": PROJECT_ID, "sid": SCENARIO_ID},
)

if gdf.empty:
    raise ValueError(f"No row for project_id={PROJECT_ID!r}, scenario_id={SCENARIO_ID!r}")

gdf
```

## 5 — Optional: Shapely geometries for center & census

Parse WKT where non-null.

```python
def _from_wkt(s):
    if s is None or (isinstance(s, float) and pd.isna(s)):
        return None
    return shapely_wkt.loads(s)

gdf["project_center_geom"] = gdf["project_center_wkt"].map(_from_wkt)
gdf["census_boundary_geom"] = gdf["census_boundary_wkt"].map(_from_wkt)

gdf[["project_id", "scenario_id", "geometry", "project_center_geom", "census_boundary_geom"]]
```

## 6 — CRS

Model defaults use SRID **4326**.

## 7 — Optional: GeoJSON export (project boundary)

Override output path with env `CIM_GEOJSON_OUT` (default `project_boundary.geojson` in the notebook working directory).

```python
from pathlib import Path
import json

if gdf.empty or gdf.geometry.isna().all():
    raise ValueError("No project_boundary geometry — cannot build GeoJSON.")

export_gdf = gdf[
    ["geometry", "project_name", "project_id", "scenario_id"]
].copy()

geojson_text = export_gdf.to_json()
geojson_obj = json.loads(geojson_text)
geojson_pretty = json.dumps(geojson_obj, indent=2, ensure_ascii=False)

out_path = Path(os.environ.get("CIM_GEOJSON_OUT", "project_boundary.geojson"))
out_path.write_text(geojson_pretty, encoding="utf-8")
print("Wrote:", out_path.resolve())
print("\nPreview (first 1200 chars):\n")
print(geojson_pretty[:1200] + ("..." if len(geojson_pretty) > 1200 else ""))
```
