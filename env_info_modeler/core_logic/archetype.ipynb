### LoD2 surfaces and CityGML

Ground is `lod0_3d_footprint_32632`. Each exterior edge, and each courtyard edge, becomes a wall extruded by `building_height_m`. The roof is the footprint lifted to the eave. A building with no height keeps the ground face only and is left out of the CityGML file.

Constructions are `{archetype}_wall`, `_roof` and `_floor`, with the U-value already on the building. A ready EuReCa heating name is written when the carrier and the heating type match one. Natural gas uses the cluster to choose condensing or traditional. The emitter is drawn with the same hash as the other archetype samples (`random_seed` 42). Oil boilers are omitted because `hasCondensation` is empty. Solid fuel, LPG and a device that heats only part of a dwelling have no single ready name.

```python
from citygml_export import assign_heating_devices, build_surfaces, write_citygml

devices = assign_heating_devices(torino_residenziale, energy_systems, random_seed=42)
surfaces = build_surfaces(torino_residenziale)
write_citygml(torino_residenziale, surfaces, devices, "torino_residenziale.gml")

surfaces.groupby("surface_type").size()
```
